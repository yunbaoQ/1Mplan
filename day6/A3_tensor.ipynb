```python
import torch
```

```python
x = torch.tensor([[1,2,3],
                [4,5,6]], dtype = torch.int64)
```

```python
print(x)
print(x.shape,x.dtype)
```

```python
s0 = x.sum(dim = 0)
s1 = x.sum(dim = 1)
print(s0, s0.shape)
print(s1, s1.shape)
```

```python
x_float = x.to(torch.float32)
print(x.dtype, x_float.dtype)
print(x_float)
```

```python
scores = torch.zeros((2, 2))
b = torch.tensor([0.5, -0.5])
print(scores.shape, b.shape)
print(scores + b)
```
