```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
x = torch.tensor(4.0, requires_grad=True)
print(x)
y = x * x + x * 3
print("x =", x.item())
print("y =", y.item())
print("y的shape =", y.shape)
print("求导前x.grad =", x.grad)
```

```python
y.backward()
print("求导后x.grad =", x.grad)
print("求导后x =", x.item())
```

```python
v = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
squared = v * v
loss = squared.sum()

print("逐元素平方 =", squared)
print("平方结果shape =", squared.shape)
print("loss =", loss.item())
print("loss的shape =", loss.shape)
```

```python
loss.backward()
print("v.grad =", v.grad)
print(v)
```

```python
u = torch.tensor(2.0, requires_grad=True)
loss1 = u * u
loss1.backward()
print("第一次 =", u.grad)
```

```python
loss2 = u * u
loss2.backward()
print("第二次，没清梯度 =", u.grad)
```

```python
u.grad = None
print("清除后 =", u.grad)

loss3 = u * u
loss3.backward()
print("清除后重新求导 =", u.grad)
```

```python
normal_result = u * u

with torch.no_grad():
    inference_result = u * u

print("正常计算 =", normal_result.requires_grad)
print("no_grad计算 =", inference_result.requires_grad)
print("u本身 =", u.requires_grad)
print("原来的u.grad =", u.grad)
print(u)
```

```python
t = inference_result
inference_result.backward()
```
