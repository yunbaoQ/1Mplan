```python
import torch
import numpy as np
```

```python
t = torch.ones(5)
print(t.device)
n = t.numpy()
print(n.device)
```

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

```python
t = t.to("cuda")
print(t.device)
print(n.device)
```

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

```python
t = t.to("cpu")
print(t.device)
print(n.device)
```

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

原来两个桥接的tensor和numpy在tensor转入cuda后桥接断开了，就算转出到cpu，桥接依旧是断开的
