```python
import torch
from torch import nn as nn
from torch.utils.data import TensorDataset, DataLoader
```

```python
torch.manual_seed(42)

x = torch.tensor([[-2.0], [-1.0], [0.0], [1.0], [2.0]],
                 dtype=torch.float32)
target = 2 * x + 1

dataset = TensorDataset(x, target)

loader = DataLoader(
    dataset,
    batch_size= 2,
    shuffle=False,
    drop_last=False,
    num_workers=0
)
```

```python
model = nn.Linear(1,1)
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(),lr = 0.05)

print(model.weight.item(),
      model.bias.item()
)

```

```python
for epoch in range(1,51):
    for batch_index, (bacth_x, batch_target) in enumerate(loader, start=1):
        optimizer.zero_grad(set_to_none=True)
        prediction = model(bacth_x)
        loss = criterion(prediction, batch_target)
        loss.backward()
        optimizer.step()

        if (epoch*batch_index)%10 == 0:
            with torch.no_grad():
                current_loss = criterion(model(bacth_x), batch_target).item()
            print(
                f"epoch={epoch}, loss={current_loss:.6f}, "
                f"w={model.weight.item():.6f}, "
                f"b={model.bias.item():.6f}"
            )
```
