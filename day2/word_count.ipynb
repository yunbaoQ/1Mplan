```python
#定义函数
def word_count(text):
    for mark in '.,!?;:':
        text = text.replace(mark, " ")
    words = text.lower().split()
    counts = {}
    for word in words:
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1    
    return counts
print(word_count('Hello,hello ai!'))
```
